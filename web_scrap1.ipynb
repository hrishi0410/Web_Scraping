```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
```

```python
url= 'https://www.scrapethissite.com/pages/forms/'
resp= requests.get(url)
print(resp.status_code)
```

```python
soup= BeautifulSoup(resp.content, 'html.parser')
print(soup.prettify())
```

```python
title= soup.title.text.strip()
print(title)
```

```python
team_player = soup.find_all('tr', class_='team')
player_names = [row.find('td', class_='name').text.strip() for row in team_player]
print(player_names)
```

```python
team_year = soup.find_all('tr', class_='team')
year = [row.find('td', class_='year').text.strip() for row in team_year]
print(year)
```

```python
wins = soup.find_all('tr', class_='team')
#win = [row.find('td', class_='pct text-success')for row in wins]
      #if row.find('td', class_='pct text-success') is not None]
win = [row.find('td', class_='pct').text.strip() for row in wins]
print(win)
```

```python
wins = soup.find_all('tr', class_='team')
player_stats = [td.text.strip() for row in wins if 'Washington' in row
                .find('td', class_='name').text for td in row.find_all('td')]
print(player_stats)
```

```python
table= soup.find_all('table', class_ ='table')
headings=  [th.text.strip() for th in table[0].find_all('th')]
print(headings)
```

```python
df = pd.DataFrame(columns= headings)
df
```

```python
column_data= table[0].find_all('tr')
```

```python
for row in column_data[1:]:
    row_data = row.find_all('td')
    individual_row_data = [data.text.strip() for data in row_data]
    print(individual_row_data)

    length = len(df)
    df.loc[length] = individual_row_data
```
